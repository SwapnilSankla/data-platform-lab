```python
from pyspark.sql import SparkSession

spark = SparkSession \
        .builder \
        .appName("narrow_transformation") \
        .getOrCreate()
```

```python
df = spark \
    .read \
    .option("header", "true") \
    .option("inferSchema", "false") \
    .csv("../../datasets/raw/2019-Nov.csv") \
    .limit(10000)
```

```python
df.show(n=1, vertical=True, truncate=False)
```

```python
df.select("event_type").distinct().show()
```
